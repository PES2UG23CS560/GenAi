```python
# Setup (Hidden)
from dotenv import load_dotenv
load_dotenv()

import getpass
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

if "GOOGLE_API_KEY" not in os.environ:
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Google API Key: ")

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
template = ChatPromptTemplate.from_template("Tell me a fun fact about {topic}.")
parser = StrOutputParser()
```

```python
# Step 1: Format inputs
prompt_value = template.invoke({"topic": "Crows"})

# Step 2: Call Model
response_obj = llm.invoke(prompt_value)

# Step 3: Parse Output
final_text = parser.invoke(response_obj)

print(final_text)
```

```python
# Define the chain once
chain = template | llm | parser

# Invoke the whole chain
print(chain.invoke({"topic": "Octopuses"}))
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI

# Assuming 'llm' is already defined from previous cells
# If not, you might need to re-run the setup cell 1Qgh-OaI3q10 or V3nwJxXT3IQb

# Create the LCEL chain in one line
movie_info_chain = (
    ChatPromptTemplate.from_template(
        "For the movie {movie_name}, which was released in {release_year}, calculate how many years ago that was from the current year. Respond *only* in the exact format: 'The movie {movie_name} was released in {release_year}, which was [YEARS_AGO] years ago.' Do not add any other text."
    )
    | llm
    | StrOutputParser()
)

# Prompt the user for input
movie_name = input("Enter the movie name: ")
release_year = int(input("Enter the release year: "))

# Invoke the chain with user-provided inputs
result = movie_info_chain.invoke({"movie_name": movie_name, "release_year": release_year})
print(result)
```
