```python
from transformers import pipeline
```

```python
summarizer = pipeline(
    "summarization",
    model="sshleifer/distilbart-cnn-12-6"
)
```

```python
article_text = """
Artificial Intelligence has seen rapid growth over the last decade.
It is now widely used in healthcare, finance, education, and transportation.
Despite its benefits, AI also introduces risks such as bias, hallucinations,
data privacy concerns, and job displacement. Governments and organizations
are working on regulations to ensure responsible AI development.
"""
```

```python
summary = summarizer(
    article_text,
    max_length=60,
    min_length=30,
    do_sample=False
)
```

```python
print("ORIGINAL TEXT:\n")
print(article_text)

print("\nSUMMARY:\n")
print(summary[0]["summary_text"])
```
